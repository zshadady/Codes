```python
import requests
from datetime import datetime
from dateutil.relativedelta import relativedelta
```

```python


url = (

    f"https://data.cityofchicago.org/resource/ajtu-isnz.json?"
    f"$where=trip_start_timestamp >= '2025-02-16T00:00:00'"
    f"AND trip_start_timestamp <= '2025-02-16T23:59:59'"
    f"&$limit=30000"
)

response = requests.get(url)
data = response.json()

len(data)
```

```python
# Getting a specific day's data
```

```python
current_datetime = datetime.now() - relativedelta(month=2)
formatted_datetime = current_datetime.strftime("%Y-%m-%d")
formatted_datetime
```

```python


url = (

    f"https://data.cityofchicago.org/resource/ajtu-isnz.json?"
    f"$where=trip_start_timestamp >= '{formatted_datetime}T00:00:00' "
    f"AND trip_start_timestamp <= '{formatted_datetime}T23:59:59' "
    f"&$limit=30000"
)

response = requests.get(url)
data = response.json()

len(data)
```
