```python
#importing the needed modules
import json
```

```python
#opening the spotify_playlist.json file in Python
file_path = r"C:\Users\win11\Documents\DataEngineer_kepzes\Codes\data\spotify_playlist.json"

with open(file_path, 'r') as file:
    data = json.load(file)
```

```python
#printing out the name of the 31th track from the playlist
counter = 1
for track in data["contents"]["items"]:
    if counter == 31:
        print("The name of the 31th track is: " + track["name"])
    counter += 1
```

```python
#printing out the number of the total playcount for all 50 tracks
counter = 1
for track in data["contents"]["items"]:
    counter += track["playCount"]
scounter = str(counter)
print("The total playcount is: " + scounter)
```

```python
#printing out the name of the track which has the lowest playcount

counter = 1
lowest_playcount = data["contents"]["items"][0]["playCount"]

for track in data["contents"]["items"]:
    current_playcount = track["playCount"]
    if current_playcount < lowest_playcount:
        lowest_name = track["name"]
        lowest_playcount = current_playcount
    counter += 1

output_text = f"The song with the lowest playcount is {lowest_name}."
print(output_text)
```
