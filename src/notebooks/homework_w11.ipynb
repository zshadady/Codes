```python
--1--list the 10 companies make the most money in descending order
SELECT 
    dc.company as COMPANY,
    ROUND(SUM(fct.trip_total),0) as TOTAL_INCOME
FROM "AwsDataCatalog"."chicago_taxi_db"."fact_taxi_trips" fct
INNER JOIN "AwsDataCatalog"."chicago_taxi_db".dim_company dc 
    ON fct.company_id = dc.company_id
group by COMPANY
order by TOTAL_INCOME desc
limit 10;
```

```python
--2--Top 10 pickup community areas with the most rides
SELECT 
    dca.community_name as PICKUP_COMMUNITY_NAME,
    COUNT(*) as COUNT_OF_RIDES
FROM "AwsDataCatalog"."chicago_taxi_db"."fact_taxi_trips" fct
LEFT JOIN "AwsDataCatalog"."chicago_taxi_db".dim_community_areas dca 
    ON fct.pickup_community_area_id = dca.area_code
group by dca.community_name
order by COUNT_OF_RIDES desc
limit 10;
```

```python
--3--Count taxi rides per day of the week
SELECT
    CASE dd.day_of_week
        WHEN 1 THEN 'Monday'
        WHEN 2 THEN 'Tuesday'
        WHEN 3 THEN 'Wednesday'
        WHEN 4 THEN 'Thursday'
        WHEN 5 THEN 'Friday'
        WHEN 6  THEN 'Saturday'
        WHEN 7 THEN 'Sunday'
    END as day_of_week,
    count(*) as trips_per_day
    
FROM
    "AwsDataCatalog"."chicago_taxi_db"."fact_taxi_trips" ftt
INNER JOIN
    "chicago_taxi_db"."dim_date" dd
    ON
    DATE(ftt.datetime_for_weather) = dd.date
GROUP BY dd.day_of_week
ORDER BY dd.day_of_week asc
```
